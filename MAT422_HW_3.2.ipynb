## 3.2 Limits and Continuity

### Definition 3.2.1: Limits of a Function
A limit describes the value that a function approaches as the input gets closer to a particular point.

For a function $f(x)$, we write $$\lim_{x\to a}f(x)=L$$

This means that as $x$ gets closer to $a$, the value of $f(x)$ gets closer to $L$.

For example, if $$f(x)=x^2,$$ then $$\lim_{x\to2}x^2=4.$$ As $x$ approaches $2$, $x^2$ approaches $4$.


```python
# A limit describes what a function approaches
def f(x):
  return x**2

# Values approaching x = 2
x_values_limit = np.array([1.5, 1.9, 1.99, 1.999, 2.001, 2.01, 2.1, 2.5])

# Calculate f(x)
y_values_limit = f(x_values_limit)

# Display the values
for x, y in zip(x_values_limit, y_values_limit):
  print(f"x={x}, f(x)={y}")
```

As $x$ gets closer to $2$ from either side, $f(x)=x^2$ gets closer to $4$. This demonstrates that $$\lim_{x\to2}x^2 = 4.$$
The function does not need to equal the limit at the surrounding points. Instead, the limit describes the value the function approaches as $x$ gets closer to the specified point.

### Definition 3.2.2: Continuous Function
A function is continuous at $a$ if its limit at $a$ is equal to its actual value at $a$. $$\lim_{x\to a}f(x)=f(a)$$. In other words, the function does not have a break or jump at that point.

For example, for $f(x)=x^2$, $$\lim_{x\to2}f(x)=4$$ and $$f(2)=4$$ Therefore, $f(x)=x^2$ is continuous at $x=2$.

```python
# Continuity means the limit equals the actual function value

x_value_continuity = 2
function_value = f(x_value_continuity)

print("f(2) =", function_value)
print("Does f(2) equal the limit 4?", np.isclose(function_value, 4))
```

### Definition 3.2.4: Extremum
An extremum is a maximum or minimum value of a function. A function has a maximum value $M$ at $z^*$ if $$f(x^*)=M$$ and $$M\geq f(x)$$ for every $x$ in the domain.

Similarly, a function has a minimum value $m$ at $z^*$ if $$f(z^*)=m$$ and $$m\leq f(x)$$ for every $x$ in the domain.

For example, if $$f(x)=-(x-2)^2+4.$$ The function reaches its maximum value at $x=2$: $$f(2)=4$$ Therefore, the maximum value is $4$.

```python
def extremum(x):
  return -(x-2)**2 + 4
x_values_extremum = np.array([0,1,2,3,4])
y_values_extremum = extremum(x_values_extremum)

for x,y in zip(x_values_extremum, y_values_extremum):
  print(f"x={x}, f(x)={y}")

# Find maximum
max_value = np.max(y_values_extremum)
max_x = x_values_extremum[np.argmax(y_values_extremum)]

print("Maximum value:", max_value)
print("Occurs at x =", max_x)
```

### Theorem 3.2.5: Extreme Value Theorem
If f is continuous on a nonempty, closed, and bounded set D, then f attains both a maxmimum and minimum on D.

```python
import numpy as np

def extreme_value_theorem(x):
    return -(x - 2)**2 + 4

x_values_evt = np.linspace(0, 4, 100)
y_values_evt = extreme_value_theorem(x_values_evt)

max_value = np.max(y_values_evt)
min_value = np.min(y_values_evt)

print("Maximum value:", max_value)
print("Minimum value:", min_value)
```

## 3.2.2 Derivatives

### Definition 3.2.6: Derivative

The derivative measures how sensitive the output of a function is to a small change in its input.

$$f'(x_0)=\lim_{h\to0}\frac{f(x_0+h) - f(x_0)}{h}$$

For example, if $$f(x)=x^2,$$ then the derivative tells us how quickly $x^2$ is changing at a particular $x$.

```python
def f(x):
  return x**2

x0 = 2
h_values = np.array([0.1, 0.01, 0.001, 0.0001])

for h in h_values:
  derivative = (f(x0 + h) - f(x0)) / h
  print(f"h={h}, f'(x0)={derivative}")
```

As h gets closer to 0, the difference quotient gets closer to 4. Therefore, the derivative of f(x) = x^2 at x = 2 is 4. This demonstrates that the derivative represents the instantaneous rate of change of the function at a specific point.

### Theorem 3.2.10 Mean Value Theorem
If $f$ is continuous on $]a,b]$ and differntiable on $(a,b)$, then there is some c between a and b such that $$f'(c)=\frac{f(b)-f(a)}{b-a}$$

Somewhere between two points, the instantaeous slope equals the overall average slope between those points.

```python
def f(x):
    return x**2

def derivative(x):
    return 2*x

a = 0
b = 5

average_slope = (f(b) - f(a)) / (b - a)

c = average_slope / 2

print("Average slope:", average_slope)
print("c:", c)
print("Derivative at c:", derivative(c))
```

For example, let $f(x)=x^2$ on the interval [0,5]. The average slope between two endpoints is 5. The Mean Value Theorem tells us that there must be some point c between 0 and 5 where the instantaneous slope is also 5. Since $f'(x)=2x$, we find c = 2.5. Therefore, at x=2.5, the instantaneous slope equals the average slope over the entire interval.

### Definition 3.2.11: Partial Derivative
$$\frac{\partial f(x_0)}{\partial x_i} = \lim_{h\to 0} \frac{f(x_0+he_i)-f(x_o)}{h}$$

A partial derivative measures how the function changes when you change one variable while holding the other variables fixed.

```python
# For f(x,y) = x^2y + 3xy^2 - 4y

def f_x(x,y):
  return (2 * x * y) + (3 * y ** 2)

def f_y(x,y):
  return (x ** 2) + (6 * x * y) - 4

x = 2
y = 3

print("f_x(2,3) =", f_x(x,y))
print("f_y(2,3) =", f_y(x,y))
```

### Definition 3.2.12: Jacobian
The Jacobian organizes all of the partial derivatives of a vector-valued function into a matrix.

Jacobian = matrix of first partial derivatives.

**Gradiant**: For a real-valued function, the Jacobian is the transpose of the gradient, so the partial derivatives appear as a row vector: $$J_f(x_0)=\nabla f(x_0)^T$$
The graident is a vector containing all of the partial derivatives.




```python
# For f(x,y) = x^2y + 3xy^2 - 4y

def f_x(x,y):
  return (2 * x * y) + (3 * y ** 2)

def f_y(x,y):
  return (x ** 2) + (6 * x * y) - 4

x = 2
y = 3

gradient = np.array([
    f_x(x,y),
    f_y(x,y)
])

jacobian = gradient.reshape(1,-1)

print("Gradient:")
print(gradient)

print("\nJacobian:")
print(jacobian)
```

The gradient contains all of the partial derivates as a column vector. For a real-valued function, the Jacobian is the transpose of the gradient, so the same partial derivatives appear as a row vector.

### Theorem 3.2.15: Chain Rule
The derivative of a composition can be found by multiplying the derivatives of the functions involved.
$$J_{gof}(x_0) = J_g(f(x_0))J_f(x_0)$$

To demonstrate the chain rule, consider the functions $$f(x)=x^2+3x+1$$ and $$g(u)=u^2.$$ Their composition is $$y(x)=g(f(x))=(x^2+3x+1)^2.$$ The derivatives of the individual functions are $$f'(x)=2x+3$$ and $$g'(u)=2u$$

```python
def f(x):
  return x**2 + 3*x + 1
def f_derivative(x):
  return 2*x + 3

def g(u):
  return u**2
def g_derivative(u):
  return 2*u

x=2

# Apply the chain rule:
# (g o f)'(x) = g'(f(x)) * f'(x)
chain_rule = g_derivative(f(x)) * f_derivative(x)

print("f(x) =", f(x))
print("f'(x) =", f_derivative(x))
print("g'(f(x)) =", g_derivative(f(x)))
print("Chain Rule derivative =", chain_rule)
```

### Definition 3.2.16: Directional Derivative
The directional derivative measures how a function changes in a particular direction $$\frac{\partial f(x_0)}{\partial v}= \lim_{h\to 0}\frac{f(x_0 + hv) - f(x_0)}{h}$$ where v is a unit vector

For example, $f(x,y) = x^2 + y^2.$ Evaluate the directional derivative at $(x_0,y_0) = (1,2)$ in the direction $v = [1,0].$

This means measuring how f changes as we move to the right along the x-axis

```python
def f_dir_der(x,y):
  return x**2 + y**2

# Point where we evaluate the directional derivative
x0 = np.array([1,2])

# Direction vector
v = np.array([1,0])

# Make sure v is a unit vector
v = v / np.linalg.norm(v)

# Values of h approaching 0
h_values = np.array([0.1,0.01,0.001,0.0001])

for h in h_values:
  x_new = x0 + h * v

  directional_derivative = (
      f_dir_der(x_new[0], x_new[1]) - f_dir_der(x0[0], x0[1])) / h

  print("h =", h, "Directional derivative =", directional_derivative)
```

```python
def f_dir_der(x,y):
  return x**2 + y**2

def gradient(x,y):
  return np.array([2*x, 2*y])

# Point where we evaluate the directional derivative
x0 = np.array([1,2])

# Direction vector
v = np.array([1,0])

# Make sure v is a unit vector
v = v / np.linalg.norm(v)

# Calculate the gradient at x0
grad = gradient(x0[0], x0[1])

# Apply the directional theorem 3.2.17
directional_derivative = np.dot(grad,v)

print("Gradient:", grad)
print("Unit direction vector:", v)
print("Directional derivative:", directional_derivative)
```

### Definition 3.2.18: Hessian
Collects the second partial derivatives into a matrix, describing how the gradient itself changes.
$$H_f(x,y)=
\begin{bmatrix}
\frac{\partial^2 f}{\partial x^2} &
\frac{\partial^2 f}{\partial x\partial y} \\
\frac{\partial^2 f}{\partial y\partial x} &
\frac{\partial^2 f}{\partial y^2}
\end{bmatrix}$$

An example is: $f(x,y) = x^2 + 3xy + y^2$

First, find the first partial derivatives:
$$f_x = 2x + 3y$$
$$f_y = 3x + 2y$$

Then, take the derivatives again:
$$f_xx = 2$$
$$f_xy = 3$$
$$f_yx = 3$$
$$f_yy = 2$$

Therefore, $H_f(x,y)=
\begin{bmatrix}
2 & 3 \\
3 & 2
\end{bmatrix}$

```python
# Second partial derivatives
def f_xx(x,y):
  return 2

def f_xy(x,y):
  return 3

def f_yx(x,y):
  return 3

def f_yy(x,y):
  return 2

x=1
y=2

hessian = np.array([
    [f_xx(x,y), f_xy(x,y)],
    [f_yx(x,y), f_yy(x,y)]
])

print("Hessian matrix:")
print(hessian)
```
