## 3.2 Limits and Continuity

### Definition 3.2.1: Limits of a Function
A limit describes the value that a function approaches as the input gets closer to a particular point.

For a function $f(x)$, we write $$\lim_{x\to a}f(x)=L$$

This means that as $x$ gets closer to $a$, the value of $f(x)$ gets closer to $L$.

For example, if $$f(x)=x^2,$$ then $$\lim_{x\to2}x^2=4.$$ As $x$ approaches $2$, $x^2$ approaches $4$.


```python
# A limit describes what a function approaches
def f(x):
  return x**2

# Values approaching x = 2
x_values_limit = np.array([1.5, 1.9, 1.99, 1.999, 2.001, 2.01, 2.1, 2.5])

# Calculate f(x)
y_values_limit = f(x_values_limit)

# Display the values
for x, y in zip(x_values_limit, y_values_limit):
  print(f"x={x}, f(x)={y}")
```

As $x$ gets closer to $2$ from either side, $f(x)=x^2$ gets closer to $4$. This demonstrates that $$\lim_{x\to2}x^2 = 4.$$
The function does not need to equal the limit at the surrounding points. Instead, the limit describes the value the function approaches as $x$ gets closer to the specified point.

### Definition 3.2.2: Continuous Function
A function is continuous at $a$ if its limit at $a$ is equal to its actual value at $a$. $$\lim_{x\to a}f(x)=f(a)$$. In other words, the function does not have a break or jump at that point.

For example, for $f(x)=x^2$, $$\lim_{x\to2}f(x)=4$$ and $$f(2)=4$$ Therefore, $f(x)=x^2$ is continuous at $x=2$.

```python
# Continuity means the limit equals the actual function value

x_value_continuity = 2
function_value = f(x_value_continuity)

print("f(2) =", function_value)
print("Does f(2) equal the limit 4?", np.isclose(function_value, 4))
```

### Definition 3.2.4: Extremum
An extremum is a maximum or minimum value of a function. A function has a maximum value $M$ at $z^*$ if $$f(x^*)=M$$ and $$M\geq f(x)$$ for every $x$ in the domain.

Similarly, a function has a minimum value $m$ at $z^*$ if $$f(z^*)=m$$ and $$m\leq f(x)$$ for every $x$ in the domain.

For example, if $$f(x)=-(x-2)^2+4.$$ The function reaches its maximum value at $x=2$: $$f(2)=4$$ Therefore, the maximum value is $4$.

```python
def extremum(x):
  return -(x-2)**2 + 4
x_values_extremum = np.array([0,1,2,3,4])
y_values_extremum = extremum(x_values_extremum)

for x,y in zip(x_values_extremum, y_values_extremum):
  print(f"x={x}, f(x)={y}")

# Find maximum
max_value = np.max(y_values_extremum)
max_x = x_values_extremum[np.argmax(y_values_extremum)]

print("Maximum value:", max_value)
print("Occurs at x =", max_x)
```

### Theorem 3.2.5: Extreme Value Theorem
If f is continuous on a nonempty, closed, and bounded set D, then f attains both a maxmimum and minimum on D.

```python
import numpy as np

def extreme_value_theorem(x):
    return -(x - 2)**2 + 4

x_values_evt = np.linspace(0, 4, 100)
y_values_evt = extreme_value_theorem(x_values_evt)

max_value = np.max(y_values_evt)
min_value = np.min(y_values_evt)

print("Maximum value:", max_value)
print("Minimum value:", min_value)
```
