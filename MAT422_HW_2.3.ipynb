## 2.3.1 Joint Probability Distributions
Joint probability is the probability of two or more events happening together. A joint probability distribution shows a probability distribution for two (or more) random variables.

### Define a Joint Probability Distribution

**Definition 2.3.1** Gives the probability that two discrete random variables take specific values at the same time. $$p(x,y) = P(X=x\,and\,Y=y)$$
All joint probabilities must be nonnegative and add up to 1.


```python
# X = result of first die
# Y = result of the second die
# Build the full 6x6 joint PMF table
joint = np.full((6,6), 1/36)
print("Joint PMF (6x6, each cell = P(X=x, Y=y)):\n", joint)

# What is the probabilities that we roll a 3 on the first die and a 5 on the second die
print("\nP(X=3, Y=5):", joint[2,4]) # die shows 3, die shows 5
print("All probabilities nonnegative:", np.all(joint >= 0))
print("Sum of all probabilities:", joint.sum())
```

### Calculate a Marginal PDF

**Marginal Distribution**: Gives the probability of one random variable without considering the value of the other variable
- To find a marginal probability, sum the joint probabilities over the other variable

**Definition 2.3.4** A marginal probability density function of X and Y, denoted by $f_x(x)$ and $f_y(y)$, respectively, are given by
$$f_x(x)=\int_{-\infty}^{\infty}f(x,y)dy \,\,for\,\, -\infty < x < \infty$$
$$f_y(y)=\int_{-\infty}^{\infty}f(x,y)dx \,\,for\,\, -\infty < y < \infty$$

```python
# X = result of first die
# Y = result of the second die

joint_PDF = np.full((6,6), 1/36)
print("Joint PMF (6x6, each cell = P(X=x, Y=y)):\n", joint_PDF)

# Probability that we roll a 3 on the first die
# We don't care what number we roll on the second die
print("\nP(X=3):", joint_PDF[2].sum())

print("All probabilities nonnegative:", np.all(joint_PDF >= 0))
print("Sum of all probabilities:", joint_PDF.sum())
```

### Calculate a Marginal PMF

**Definition 2.3.2** The marginal PMF gives the probability of one random variable while ignoring the other variables
- For X: sum over all possible values of Y $$p_X(x)=\Sigma_y p(x,y)$$
- For Y: sum over all possible values of X $$p_Y(y)=\Sigma_x p(x,y)$$

```python
## Definition 2.3.2: marginal PMF, to get the probability of X = x, add the joint probabilities for that x across all possible Y's and vice versa for x
# Example:

p_X = joint.sum(axis = 1) # sum across each row
p_Y = joint.sum(axis = 0) # sum across each column

print("Marginal X:", p_X)
print("Marginal Y:", p_Y)
print("Each marginal should be 1/6:", np.allclose(p_X, 1/6), np.allclose(p_Y, 1/6))
```

For discrete variables:

If the above is not satisfied for all (x,y), then X and Y are said to be dependent

## 2.3.2 Correlation and Dependence
When two random variables X and Y are not independent, it is frequently of interest to assess how strongly they are related to one another.



### Calculate Covariance

**Definition 2.3.9** Covariance measures how two random variables vary together
$$Cov(X,Y) = E[(X-\mu x)(Y-\mu y)]$$
- Positive covariance: X and Y tend to move in the same direction
- Negative covariance: X and Y tend to move in opposite directions
- Covariance near 0: Little/no linear relationship

```python
x_hours_studied = np.array([1,2,3,4,5])
y_exam_score = np.array([60,65,70,80,85])

# population covariance
covariance = np.cov(x_hours_studied, y_exam_score, ddof=0)[0,1]

print(covariance)
```

### Calculate Correlation

**Definition 2.3.10** The correlation coefficient of X and Y, denoted by $$Corr(X,Y), \rho_{x,y}$$ or just $\rho$, is defined by $$\rho_{x,y}={Cov(X,Y)}/({\sigma_x * \sigma_y})$$
- Correlation is always between -1 and 1
- $\rho = 1$: perfect positive linear relationship
- $\rho = -1$: perfect negative linear relationship
- $\rho = 0$: no linear relationship
If X and Y are independent:
- X and Y independent: $\rho = 0$, but $\rho = 0$ does not necessarily mean X and Y are independent. Correlation only measures linear relationships

```python
# population correlation
correlation = np.corrcoef(x_hours_studied,y_exam_score)[0,1]
print("Correlation:", correlation)
```

### Calculate Sample Covariance
Estimates the covariance between two variables using sample data.
$$s_{xy}=1/(n-1)\,\Sigma_{i=1}^{n}(x_i - \bar{x})(y_i - \bar{y})$$


```python
# as X increases, Y increases too, so the covariance is positive

sample_covariance = np.cov(x_hours_studied,y_exam_score)[0,1]
print(sample_covariance)
```

### Calculate Sample Correlation

Measures the strength and direction of the linear relationship between two variables in a sample of data.
$$r=s_{xy}/s_xs_y$$ where:
- $s_{xy}$ = sample covariance
- $s_x$ = sample standard deviation of X
- $s_y$ = sample standard deviation of Y

```python
# sample correlation
r = np.corrcoef(x_hours_studied,y_exam_score)[0,1]
print("Sample correlation:", r)

# negative correlation example
x_1 = np.array([1,2,3,4,5])
y_1 = np.array([10,9,8,7,6])
neg_cov = np.cov(x_1,y_1)[0,1]
print("Negative sample correlation example:", neg_cov)
```

## 2.3.3 Random Samples
A random sample is a group of observations that are independent and come from the same population/distribution.

A random sample of size n consists of random variables X1,...,Xn that:
1. Are independent of each other
2. Have the same probability distribution

### Define a Random Sample

```python
population = [1,2,3,4,5,6] # die outcomes = the population
pop_mean = np.mean(population)
pop_std = np.std(population)
print("Population mean:", pop_mean, " Population std:", pop_std)

n = 30
sample = np.random.choice(population, size=n, replace=True)
print("Random sample of size", n, ":", sample)
```

### Calculate the Sample Mean

Sample Mean
- The sample mean is the average of the observations:
$$\bar{X} = (X_1+...+X_n)/n$$ The sample mean is used to estimate the population mean $\mu$

As the sample size n gets larger: $\sigma_{\bar{X}} = (\sigma)/\sqrt{n}$ gets smaller

The sample mean is centered around the population mean, and its variability gets smaller as the sample size increases.
$$E(\bar{X}) = \mu$$

$$V(\bar{X}) = \frac{\sigma^2}{n}$$

$$\sigma_{\bar{X}} = \frac{\sigma}{\sqrt{n}}$$

```python
# If we take lots of different samples from the same population, their sample means will vary.

# Take one sample
sample = np.random.choice(population, size = n, replace=True)

# Find its mean
sample_mean = np.mean(sample)

print("Sample:", sample)
print("Sample mean:", sample_mean)

# demonstrate sigma_xbar = sigma/sqrt(n)
sample_means = []
for i in range(10000):
  sample = np.random.choice(population, size=n, replace=True)
  sample_mean = np.mean(sample)
  sample_means.append(sample_mean)

print("Standard deviation of sample means:", np.std(sample_means))
print("sigma/sqrt(n):", pop_std/np.sqrt(n))
```

### Calculate the Sample Total

Sample Total: The sample total is $$T_0=X_1+...+X_n$$ its properties are: $$E(T_0)=n\mu$$ $$V(T_0)=n\sigma^2$$ $$\sigma_{T_0} = \sqrt{n}\sigma$$

```python
sample_total = np.sum(sample)
print("Sample total:", sample_total, " (E(T) = n*mu =", n*pop_mean, ")")

totals = []
for i in range(10000):
  sample = np.random.choice(population, size=n, replace=True)
  sample_total = np.sum(sample)
  totals.append(sample_total)

print("Mean of totals:", np.mean(totals), " vs n*mu:", n*pop_mean)
print("Std of totals:", np.std(totals), " vs sqrt(n)*sigma:", np.sqrt(n)*pop_std)
```

### Demonstrate the Central Limit Theorem

**2.3.3.2 Central Limit Theorem**: Even if the original population is not normally distributed, the distribution of the sample mean becomes approximately normal when the sample size is sufficiently large.

For a large $n$:

$$\bar{X} \approx N\left(\mu,\frac{\sigma^2}{n}\right)$$

- Mean of $\bar{X}$: $\mu$
- Variance of $\bar{X}$: $\frac{\sigma^2}{n}$
- Standard deviation of $\bar{X}$: $\frac{\sigma}{\sqrt{n}}$

The larger the sample size $n$, the better the normal approximation.

```python
# Compare small and large sample sizes
# Larger samples give sample means that are more tightly clustered

small_n = 5
large_n = 50

# store sample means for n = 5
means_small = []

for i in range(5000):
  sample = np.random.choice(population, size=small_n, replace=True)
  sample_mean = np.mean(sample)
  means_small.append(sample_mean)

# store sample means for n = 50
means_large = []

for i in range(5000):
  sample = np.random.choice(population, size=large_n, replace=True)
  sample_mean = np.mean(sample)
  means_large.append(sample_mean)

print("n=5  -> standard deviation of means:", np.std(means_small), " standard deviation of X bar:", pop_std/np.sqrt(small_n))
print("n=50 -> standard deviation of means:", np.std(means_large), " standard deviation of X bar:", pop_std/np.sqrt(large_n))
```
