## 2.4.1. MLE for random samples

Maximum likelihood estimation is a method used to estimate unknown parameters of a probability distribution. The idea is to choose the parameter value that makes the observed data most likely to have occurred.

GIven the observed data $x_1, x_2,..., x_n$, the likelihood function is $$L(\theta_1,...,\theta_m) = f(x_1,x_2,...,x_n;\theta_1,...,\theta_m)$$

The MLE is the value of the parameter(s) that maximizes this likelihood: $$\hatθ = \underset{\theta}{\operatorname{argmax}}\,L(\theta)$$

If we observe several coin flips and want to estimate the probability p of getting heads, MLE finds the value of p that makes our observed sequence of heads and tails most likely.

```python
# If we observe H, H, T, H, T (random sample of coin flips)
X_rand = ['H', 'H', 'T', 'H', 'T']

# Count the number of heads
head_count = 0
length = len(X_rand)

for flip in X_rand:
  if flip == 'H':
    head_count += 1

# Calculate the MLE for p
p_hat = head_count / length

print("Estimated probability of heads:", p_hat)

# Likelihood function:
# L(p) = p^3 * (1-p)^2
def likelihood(p):
  return p**3 * (1-p)**2

# Calculate the likelihood at the estimated p
L_hat = likelihood(p_hat)

print("Likelihood at p =", p_hat, ":", L_hat)
```

### MLE for a Normal Distribution
A normal distribution has two unknown parameters: $\mu = mean$ and $\sigma^2 = variance$. We ask what values of $\mu$ and $\sigma^2$ makes the observed data most likely under a normal distribution.

For a random sample $(X_1,X_2,\ldots,X_n)$ from a normal distribution, the maximum likelihood estimates are $\hat{\mu}=\bar{X}$ and $\hat{\sigma}^2=\frac{1}{n}\sum_{i=1}^{n}(X_i-\bar{X})^2$

Thus, the MLE of the mean is the sample mean, and the MLE of the variance is the average squared distance of the observations from the sample mean.

```python
# Observed sample
X = np.array([2,4,5,6,8])

# MLE for the mean
mu_hat = np.mean(X)
mu_calculation = np.sum(X) / len(X)

# MLE for the variance
sigma2_hat = np.mean((X - mu_hat)**2)
sigma2_calculation = np.sum((X - mu_calculation)**2) / len(X)

# MLE for the standard deviation
sigma_hat = np.sqrt(sigma2_hat)

print("MLE for mean:", mu_hat)
print("Does mu_hat and mu_calculation equal:", np.allclose(mu_hat, mu_calculation))
print()
print("MLE for variance:", sigma2_hat)
print("Does sigma2_hat and sigma2_calculation equal:", np.allclose(sigma2_hat, sigma2_calculation))
print()
print("MLE for standard deviation", sigma_hat)
```

## 2.4.2. Linear regression

Linear regression finds a line that best fits a set of data points. The predicted value of (y) is given by $$\hat{y}i = \beta_0 + \beta_1x{i1}+\cdots+\beta_px_{ip}$$ where the $\beta$'s are the coefficients of the regression model.

From a probabilistic perspective, we assume that the observed y-values are normally distributed around the predicted values $\hat{y}_i$. In other words, the regression line represents the mean of the normal distribution for each ${x_i}$.

Because the observations are assumed to be independent, the likelihood of observing all of the data is the product of the individual normal probability densities:

$$\prod_{i=1}^{n} \frac{1}{\sigma\sqrt{2\pi}} e^{-\frac{(y_i-\hat{y}_i)^2}{2\sigma^2}}$$

Maximum likelihood estimation chooses the regression coefficients that make the observed data most likely. For a normal-error model, maximizing the likelihood is equivalent to minimizing the sum of squared errors:

$$\sum_{i=1}^{n}(y_i-\hat{y}_i)^2$$ Thus, the usual least-squares approach to linear regression can also be understood as a maximum likelihood estimation model.

```python
# We want to find a line that predicts exam score based on hours studied
# The regression model assumes that scores are normally distributed around the predicted value of the regression line

# Observed data
x_reg = np.array([1,2,3,4,5])
y_reg = np.array([55,60,65,72,78])

# Fit a linear regression line
beta_1, beta_0 = np.polyfit(x_reg,y_reg,1)

# Predicted values
y_hat = beta_0 + beta_1 * x_reg

print("Intercept (β0):", beta_0)
print("Slope (β1):", beta_1)
```

### Linear Regression from a Probabilistic Perspective
In linear regression, the predicted value is not expected to equal the actual value exactly. Instead, we assume there is some random error:

$y_i=\hat{y}_i+\epsilon_i$
where the error is assumed to follow a normal distribution:
$\epsilon\sim N(0,\sigma^2)$

Therefore, each observed y_i is normally distributed around its predicted value $\hat{y}_i$: $$y_i\sim N(\hat{y}_i,\sigma^2)$$
This means that the regression line gives the mean of the normal distribution for each $x_i$.

Because the observations are independent, we multiply the probability of observing each y_i to obtain the likelihood: $$\prod_{i=1}^{n} \frac{1}{\sigma\sqrt{2\pi}} e^{-\frac{(y_i-\hat{y}_i)^2}{2\sigma^2}}$$

MLE chooses the regression coefficients $\beta$ that maximizes this likelihood. When we simplify the likelihood, maximizing it is equivalent to minimizing: $\sum_{i=1}^{n}(y_i-\hat{y}_i)^2$

This is exactly the least-squares criterion used in linear regression.

Therefore, when the errors are normall distributed, maximum likelihood estimation and least-squares regression produce the same regression coefficients.

```python
# Find the least-squares regression line
beta_1, beta_0 = np.polyfit(x_reg,y_reg,1)
# Predicted values
y_hat = beta_0 + beta_1 * x_reg

# Calculate residuals
residuals = y_reg - y_hat

# Calculate sum of squared errors
SSE = np.sum(residuals**2)

print("Intercept:", beta_0)
print("Slope:", beta_1)
print("Predicted values:", y_hat)
print("Residuals:", residuals)
print("Sum of squared errors:", SSE)
```
