## 1.4.2 Low-Rank Matrix Approximations

Low-rank approximation is a mathematical method used to replace a large, complex data matrix with a smaller, simplified matrix of lower rank while keeping the most important information. This section demonstrates the core definitions behind low-rank matrix approximations: determining how an SVD breaks a matrix into components, learning how low-rank approx. means keaping only the most important components, and checking to see that the approximation error is the next singular value.


### SVD breaks a matrix into components
$$
A = Σ σ_iu_iv_i^T
$$


### Low-rank approximation means keeping only the most important components

**Definition 1.4.3** Truncated SVD
$$
A_k=\sum_{i=1}^{k} \sigma_i u_i v_i^T
$$

### Induced 2-norm measures the maximum stretching

**Definition 1.4.3** The 2 norm measures the maximum amount that matrix A can stretch a vector (how big the error between two matrices is)

### The approximation error is the next singular value

**Lemma 1.4.4** The error you get from keeping the first k components is exactly the next singular value
 $$\|A-A_k\|_2=\sigma_{k+1}$$

### Truncated SVD gives the best possible low-rank approximation

**Theorem 1.4.5** Eckart-Young Mirsky $||A - A_k||_2 <= ||A - B||_2$ for any B with: rank(B) <= k

No other rank-k approximation can do better than $A_k$
