## 2.2.1 Probability Axioms
Probability axioms define how probabilities are assigned to events in a sample space. This section demonstrates sample space, simple and compound events, probability distributions, and the basic properties that probabilities must satisfy

### Define the experiment and sample space
**Definition 2.2.1**: The sample space of an experiment, denoted by S, is the set of all possible outcomes of that experiment. In probability, we are interested in all the various collections of outcomes from S.

```python
# Experiment roll a fair six-sided die once

outcomes = [1,2,3,4,5,6]
sample_space = set(outcomes)
print("Sample space S:", sample_space)
```

### Demonstrate a simple event
**Definition 2.2.2**: An event is any collection (subset) of outcomes contained in the sample space S. An event is simple if it consists of exactly one outcome and compound if it consists of more than one outcome.


```python
# Event A = rolling a 3
A = {3}
P_A = len(A) / len(outcomes)
print("Simple event A (rolling a 3):", A)
print("P(A):", P_A)
```

### Demonstrate a compound event

```python
# Event B = rolling an even number
B = {2,4,6}
P_B = len(B) / len(outcomes)
print("Compound event B (rolling an even number):", B)
print("P(B):", P_B)
```

### Demonstrate the probability distribution
**Definition 2.2.3**: Given an experiment and a sample space S, the probability distribution is a function which assigns to each event A a number P(A), called the probability of the event A, which will give a precise measure of the chance that A will occur. It should satisfy the basic properties of probability.


```python
pmf = {}
for x in outcomes:
  pmf[x] = 1 / len(outcomes)

print("Probability distribution (fair die):", pmf)

total = 0
for value in pmf.values():
  total += value

print("Sum of all probabilities (should be 1):", total)
```

### Verify the basic probability properties
**Property 1:** probabilities lie between 0 and 1

**Property 2:** the probability of the entire sample space is 1

**Property 3:** probabilities of disjoint events add: $P(A⋃B) = P(A)+P(B)$

```python
# Property 1
print("Property 1 (0 <= P(B) <= 1):", 0 <= P_B <= 1)
# Property 2
p_sample_space = len(sample_space) / len(outcomes)
print("Property 2 (P(S) = 1):", p_sample_space == 1)
# Property 3
odd = {1,3,5}
even = {2,4,6}
p_odd = len(odd) / len(outcomes)
p_even = len(even) / len(outcomes)
p_union = len(odd | even) / len(outcomes)

print("Property 3 (disjoint probabilities add):", math.isclose(p_odd + p_even, p_union))
```

## 2.2.2 Conditional Probability
Conditional probability is the likelihood of an event, given that another event already happened.

### Conditional Probability: B changes the sample space
**Definition 2.2.5** For events A and B with P(B) > 0, the conditional probability of A given B is P(A|B) = P(A∩B) / P(B). Knowing B happened shrinks the relevant sample space from S down to just B.

```python
S = set(outcomes)
A2 = {2,4,6}
B2 = {4,5,6}

intersection = A2 & B2

print("Before knowing B: S=", S)
print("After knowing B: relevant space=", B2)
print("A ∩ B:", intersection)

P_intersection = len(intersection) / len(S)
P_B2 = len(B2) / len(S)
P_A_given_B = len(intersection) / len(B2)
print("P(A|B):", P_A_given_B)
```

### Multiplication Rule
Conditional probability gives rise to the multiplication rule: $P(A\cap B) = P(A|B) * P(B)$

```python
lhs = P_A_given_B * P_B2
print("P(A|B) * P(B):", lhs)
print("(A∩B):", P_intersection)
print("Multiplication rule holds:", math.isclose(lhs, P_intersection))
```

### Demonstrate independence
**Definition 2.2.6**: A and B are independent if $P(A|B) = P(A)$, and dependent otherwise

```python
# A and B above are dependent: knowing B happened changes the probability of A. Then, two separate dice rolles, which are really independent, since one roll can't affect the other.
P_A2 = len(A2) / len(S)
print("P(A):", P_A2)
print("P(A|B):", P_A_given_B)
print("Independence:", math.isclose(P_A2, P_A_given_B))

# Two separate dice rolls: A = first roll even, B = second roll odd
# By definition P(A and B) = P(A) * P(B)
A_indep = {2,4,6}
B_indep = {1,3,5}
P_A_indep = len(A_indep) / len(S)
P_B_indep = len(B_indep) / len(S)
P_intersection_indep = P_A_indep * P_B_indep

print("\nSeparate dice: P(A):", P_A_indep, " P(B):", P_B_indep)
print("P(A) * P(B):", P_A_indep * P_B_indep)
print("P(A ∩ B) (independent trials):", P_intersection_indep)
```

### Demonstrate mutually independent events
**Definition 2.2.7** Events $A_1,...,A_n$ are mutually independent if, for every subset of them, the probability of their intersection equals the product of their individual probabilities.

```python
# Roll a die three times: A = first roll even
# B = second roll even
# C = third roll even

p_even_single = 0.5 # P(A) = P(B) = P(C) for a single fair die roll
P_AB = p_even_single * p_even_single
P_AC = p_even_single * p_even_single
P_BC = p_even_single * p_even_single
P_ABC = p_even_single ** 3

print("P(A∩B):", P_AB)
print("P(A∩C):", P_AC)
print("P(B∩C):", P_BC)
print("P(A∩B∩C):", P_ABC)
```

## 2.2.3 Discrete Random Variables
**Definition 2.2.8** A random variable is a rule assigning a numerical value to each outcome in the sample space. For a die, X = the number rolled (X just represents the outcome as a number)

**Definition 2.2.9** A discrete random variable has a finite (or countably infinite) set of possible values

### Define a random variable

```python
# X = the number rolled on a fair six-sided die
X_values = outcomes
print("Random variable X: the number rolled")
```

### Show its possible values

```python
print("Possible values of X:", X_values)
```

### Build its PMF
**Definition 2.2.10** The probability mass function gives the probability that X is exactly equal to a value: p(x) = P(X=x)

```python
pmf2 = {}
for x in X_values:
  pmf2[x] = 1 / len(X_values)

print("PMF:", pmf2)
```

### Build its CDF
**Definition 2.2.11** The cumulative distribution function (CDF) gives the probability that X is less than or equal to a value: F(x) = P(X ≤ x). For example, F(3) should be 3/6 = 0.5, since {1,2,3} are the outcomes at or below 3.

```python
cdf = {}
total2 = 0
for x in sorted(X_values):
  total2 += pmf2[x]
  cdf[x] = total2

print("CDF:", cdf)
print("F(3) == 0.5:", math.isclose(cdf[3], 0.5))
```

### Calculate expected value
**Definition 2.2.14** The expected value(mean) is the weighted average of all possible values of $X: E(X) = \Sigma x * p(x)$

```python
expected_value = 0
for x in outcomes:
  x_probability = 1 / len(outcomes)
  expected_value += x * x_probability

print("Expected value (mean):", expected_value)
```

### Calculate variance
**Definition 2.2.17** The variance measures how spread out X's values are from its mean: $V(X) = \Sigma (x-\mu )^2 * p(x)$

```python
variance = 0
for x in outcomes:
  x_probability = 1 / len(outcomes)
  variance += (x - expected_value)**2 * x_probability

print("Variance:", variance)
```

### Calculate standard deviation
Standard deviation is the square root of variance: $\sigma = \sqrt V(X)$

```python
standard_deviation = math.sqrt(variance)
print("Standard Deviation:", standard_deviation)
```

### Demonstrate Bernoulli and Binomial
**Example 2.2.12** A bernoulli random variable has exactly two outcomes, $X \in {0, 1}.$ Let X = 1 if a roll is even (success), X = 0 if odd (failure); $P(X=1)=3/6=1/2$

A binomial random variable counts the number of successes across n independent Bernoulli trials. Rolling the die 5 times and counting even gives a binomial random variable

```python
n = 5 # number of trials
p = 0.5 # probability of success (rolling even) on each trial

outcomes = [1,2,3,4,5,6]

# simulate 5 rolls, counting on how many are even
for roll in outcomes:
  if roll % 2 == 0:
    X = 1
  else:
    X = 0
  print(f"Roll: {roll}, X: {X}")

successes = 0
for i in range(n):
  roll = random.randint(1,6)
  if roll % 2 == 0:
    successes += 1

# binomial formula: probability of exactly k successes out of n trials
k = 3
mu = 4

probability_binomial = math.comb(n, k) * p**k * (1-p)**(n-k)
print("Binomial probability:", probability_binomial)
```

### Demonstrate Poisson
**Example 2.2.13**: A Poisson distribution describes the probability of a certain number of events occuring in a fixed interval: $$p(x:\mu ) = (e^(-\mu)\mu^x)/x!$$ where $\mu$ is the average number of events.

```python
# Example: number of customers arriving at a store in one hour, with an average of 4 per hour

mu = 4 # average number of arrivals per hour
x = 3 # probability of exactly 3 arrivals

probability_poisson = (math.exp(-mu) * mu**x) / math.factorial(x)

print("P(X = 3 arrivals | mu = 4):", probability_poisson)
```

## 2.2.4 Continuous Random Variables

A continuous random variable can take any value within an interval (e.g. height could be 170.1, 170.12, 170.123...).

### Demonstrate PDF
**Definition 2.2.20** A continuous random variable has a probability density function f(x). Probability over an interval is the area under the curve: $P(a <=X<=b)= \int_{a}^{b} f(x)dx.$ Two requirements: $f(x) >= 0$, and the total area under f equals 1.

```python
# Example: X = height of a randomly selected student, X ~ N(170,8^2)
mu_h, sigma_h = 170, 8 # mean height 170cm, standard deviation 8cm
rv = norm(mu_h, sigma_h)

# probability that height falls between 162 and 178 (within 1 SD of mean)
prob_between = rv.cdf(178) - rv.cdf(162)
print("Probability between 162 and 178cm:", prob_between)
```

### Demonstrate Expected value
**Definition 2.2.21** For a continuous random variable, $E(X) = \int x * f(x)dx$

```python
expected_height = rv.mean()
print("E(X) (expected height):", expected_height)
```

### Demonstrate variance and standard deviation
**Definition 2.2.22** $V(X) = \int (x-\mu)^2 * f(x)dx$

```python
height_variance = rv.var()
height_std = rv.std()

print("Variance of height:", height_variance)
print("Standard deviation of height:", height_std)
```

### Demonstrate exponential distribution
**Definition 2.2.24** The exponential distribution is commonly used for waiting times, controlled by a rate parameter $\lambda$. With $lambda = 1$, the mean and variance both equal 1.

```python
lam = 1 # rate peramater
exp_rv = expon(scale=1/lam)

print("Exponential mean:", exp_rv.mean())
print("Exponential variance:", exp_rv.var())
```

### Demonstrate normal distribution

**Definition 2.2.25** The normal distribution is a bell-shaped continuous distribution controlled by mean $\mu$ (center) and a standard deviation $\sigma$ (spread).

```python
print("Height distribution: N(mu=", mu_h, ", sigma=", sigma_h, ")")
print("PDF value at the mean (should be the peak):", rv.pdf(mu_h))
```

### Demonstrate standard normal
**Definition 2.2.26** The standard normal distribution is a normal distribution with $\mu$ = 0 and $\sigma$ = 1, denoted Z.

```python
std_normal = norm(0,1)
print("Standard normal mean:", std_normal.mean())
print("Standard normal std:", std_normal.std())
```

### Demonstrate Z-score
**Proposition 2.2.27** Any normal variable X can be converted to a standard normal variable Z via $Z = (X-\mu)/\sigma$. This tells you how many standard deviations X is from the mean

```python
# example: converting a height of 182 cm
x_height = 182
z = (x_height - mu_h) / sigma_h

print("Z-score for height 182cm:", z)
print("(This height is", z, "standard deviations above the mean)")

print("P(Z <= z) using standard normal:", std_normal.cdf(z))
print("Matches P(X <= 182) using original normal:", rv.cdf(182))
```
