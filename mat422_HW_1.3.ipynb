## 1.3.2 Least-Squares Problem

The least-squares problem finds the x that minimizes the difference between Ax and b. This problem occurs when there is a system Ax = b that doesn't have an exact solution, usually because there are more equations than unknowns.$$x^* = \arg\min_x\|Ax - b\|_2$$

### Find the x that makes the error between Ax and b as small as possible

**Theorem 1.3.2** If $A = QR$, the least-squares solution x* can be found by solving $$Rx^* = Q^Tb$$ Since R is upper triangular, x* can be found using back substitution.

## 1.3.3 Linear Regression
Linear Regression is a method used to predict an output variable from one or more input variables by finding a line that minimizes the difference between the predicted and actual values.

The basic prediction is: $$\hat{y}_i = \beta_0 + \beta_1x_{i1} + ... + \beta_dx_{id}$$ The goal is to choose the coefficients, $\beta_0, \beta_1, ..., \beta_d$ so that the predictions are as close as possible to the actual values.

### Express linear regression as $Ax ≈ b$

For linear regression, we can write the predictions as:
$$\hat{y} = X\beta$$
where X contains the input data and $\beta$ contains the coefficients. Because the predictions will not perfectly match the actual values: $$y \approx X\beta$$ This makes linear regression a least-squares problem.

### Solve the regression problem using least squares

We find the coefficients $\beta^*$ that minimize the error:

$$\beta^* = \arg\min_\beta \|X\beta-y\|_2$$
