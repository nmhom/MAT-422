```python
import numpy as np
from scipy.linalg import null_space

X = np.array([
    [1, 1],
    [1, 2],
    [1, 3]
])

y = np.array([
    2,
    2.9,
    4.2
])

A = X
b = y
```

## 1.3.1 QR Decomposition
QR factorization decomposes a matrix A into A = QR, where Q has orthonormal columns and R is upper triangular. This section demonstrates the core components of QR Decomposition: verifying that QR reconstructs A, and verifying that Q is orthonormal and R is upper triangular.

### Verify that QR reconstructs A

```python
Q, R = np.linalg.qr(A)

print("QR reconstructs A:", np.allclose(A, Q @ R))
```

### Demonstrate that Q is orthonormal and R is upper triangular

```python
print("\nQ is orthonormal:", np.allclose(Q.T @ Q, np.eye(Q.shape[1])))

print("\nR is upper triangular:", np.allclose(R, np.triu(R)))
```

## 1.3.2 Least-Squares Problem

The least-squares problem finds the x that minimizes the difference between Ax and b. This problem occurs when there is a system Ax = b that doesn't have an exact solution, usually because there are more equations than unknowns.$$x^* = \arg\min_x\|Ax - b\|_2$$

### Find the x that makes the error between Ax and b as small as possible

**Theorem 1.3.2** If $A = QR$, the least-squares solution x* can be found by solving $$Rx^* = Q^Tb$$ Since R is upper triangular, x* can be found using back substitution.

```python
x_qr = np.linalg.solve(R, Q.T @ b)

print("x* from QR (Rx=Q^Tb):", x_qr)
```

### Solve the normal equations $A^TAx = A^Tb$
**Theorem 1.3.1** The solution to a least-squares problem satisfies $A^TAx = A^Tb$. Solving this should give the same x as the QR approach above.

```python
x_normal = np.linalg.solve(A.T @ A, A.T @ b)
print("x* from normal equations (A^TAx=A^Tb):", x_normal)

left = A.T @ A @ x_normal
right = A.T @ b

print("A^TAx == A^Tb:", np.allclose(left, right))
print("Matches x* from QR:", np.allclose(x_normal, x_qr))
```

### Demonstrate the residual/error
$r = b - Ax^*$ where r is the residual vector, and $$\|r\|_2 = \|b - Ax^*\|_2$$ is the size of the residual/error.

```python
residual = A @ x_normal - b
error = np.linalg.norm(residual)

print("Residual vector:", residual)
print("Size of error:", error)
```

### Demonstrate that the least-squares residual is orthogonal to the columns of A.
We can show $A^Tr = 0$ because $r = b - Ax*$ and therefore $A^T(b-Ax^*) = 0$. This is also directly connected to the normal equations: $A^Tb-A^TAx^* = 0$ which gives $A^TAx^* = A^Tb$.

```python
x_star, residuals_sum_sq, rank, singular_values = np.linalg.lstsq(A, b, rcond=None)
residual_vector = A @ x_star - b

print("\nLeast-squares solution:", x_star)
print("\nResidual vector r = Ax* - b:", residual_vector)

print("\nSum of squared residuals (lstsq output):", residuals_sum_sq)
print("sum(r^2) computed directly:", np.sum(residual_vector**2))
print("Do these match:", np.allclose(np.sum(residual_vector**2), residuals_sum_sq))

print("\nIs the residual orthogonal to the A cols:", np.allclose(A.T @ residual_vector, 0))
```

## 1.3.3 Linear Regression
Linear Regression is a method used to predict an output variable from one or more input variables by finding a line that minimizes the difference between the predicted and actual values.

The basic prediction is: $$\hat{y}_i = \beta_0 + \beta_1x_{i1} + ... + \beta_dx_{id}$$ The goal is to choose the coefficients, $\beta_0, \beta_1, ..., \beta_d$ so that the predictions are as close as possible to the actual values.

### Express linear regression as $Ax ≈ b$

For linear regression, we can write the predictions as:
$$\hat{y} = X\beta$$
where X contains the input data and $\beta$ contains the coefficients. Because the predictions will not perfectly match the actual values: $$y \approx X\beta$$ This makes linear regression a least-squares problem.

### Solve the regression problem using least squares

We find the coefficients $\beta^*$ that minimize the error:

$$\beta^* = \arg\min_\beta \|X\beta-y\|_2$$

```python
beta = np.linalg.lstsq(X, y, rcond=None)[0]

print("Fitted coefficients (intercept, slope):", beta)
print("Predicted y:", X @ beta)
print("Actual y:", y)
```
