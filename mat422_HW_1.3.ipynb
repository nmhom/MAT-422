## 1.3.2 Least-Squares Problem

The least-squares problem finds the x that minimizes the difference between Ax and b. This problem occurs when there is a system Ax = b that doesn't have an exact solution, usually because there are more equations than unknowns.$$x* = arg min ||Ax - b||_2 $$

### Find the x that makes the error between Ax and b as small as possible
